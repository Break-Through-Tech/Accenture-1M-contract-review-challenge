```python
import json
import os
print(os.getcwd())
```

```python
with open('../data/cuad/train_separate_questions.json') as f:
    data = json.load(f)

print(type(data))
print(data.keys())
```

```python
contracts = data['data']
print(type(contracts))
print(len(contracts))
```

```python
first_contract = contracts[0]
print(type(first_contract))
print(first_contract.keys())
```

```python
print(first_contract['title'])
paragraphs = first_contract['paragraphs']
print(type(paragraphs))
print(len(paragraphs))
```

```python
first_paragraph = paragraphs[0]
print(first_paragraph.keys())
```

```python
context = first_paragraph['context']
print(type(context))
print(len(context))
print(context[:300])
```

```python
qas = first_paragraph['qas']
print(type(qas))
print(len(qas))
print(qas[0])
```

```python
for qa in qas[:10]:
    question = qa['question']
    category = question.split('"')[1]
    if qa['answers']:
        answer_text = qa['answers'][0]['text']
    else:
        answer_text = None
    print(category, '->', answer_text)
```

```python
for qa in qas[10:25]:
    question = qa['question']
    category = question.split('"')[1]
    if qa['answers']:
        answer_text = qa['answers'][0]['text']
    else:
        answer_text = None
    print(category, '->', answer_text)
```
